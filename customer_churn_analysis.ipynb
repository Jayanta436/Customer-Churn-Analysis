```python
plt.figure(figsize=(3,3))

ct = pd.crosstab(df['SeniorCitizen'], df['Churn']) #crosstab to find frequency distribution

# Convert to percentages within each SeniorCitizen group
pct = ct.div(ct.sum(axis=1), axis=0) * 100
'''
calculates the sum of each row in the ct DataFrame (i.e., the total number of customers for each SeniorCitizen category).
ct.div(..., axis=0) divides each element in ct by the corresponding row sum. axis=0 ensures that division happens row-wise.
* 100 converts the proportions into percentages.
The result, pct, is a DataFrame showing the percentage of Churned and Stayed customers for 'No' and 'Yes' SeniorCitizen categories.
'''
# Plot stacked bar chart
ax = pct.plot(kind='bar',
              stacked=True,
              figsize=(5,4))

# Add percentage labels
for container in ax.containers:
    labels = [f'{v:.1f}%' if v > 0 else '' for v in container.datavalues]
    ax.bar_label(container, labels=labels, label_type='center')

plt.title('Churn by Senior Citizen')
plt.ylabel('Percentage')
plt.xlabel('Senior Citizen')
plt.legend(title='Churn')
plt.show()
```

```python
cols=['PhoneService', 'MultipleLines',
       'InternetService', 'OnlineSecurity', 'OnlineBackup',
       'DeviceProtection', 'TechSupport', 'StreamingTV',
       'StreamingMovies', 'Contract', 'PaperlessBilling', 'PaymentMethod']
ncol=4
nrow=(len(cols)+ncol-1)//ncol
#plt.figure(figsize=(15,10))
fig,axes=plt.subplots(nrow,ncol, figsize=(20, nrow*4))
plt.tight_layout()
axes=axes.flatten()
for i,col in enumerate(cols):
  sns.countplot(x=col,data=df,ax=axes[i],hue=df["Churn"])
  axes[i].set_title(f'Countplot of {col}')
  axes[i].set_xlabel(col)
  axes[i].set_ylabel('Count')
for j in range(i+1,len(axes)):
  fig.delaxes(axes[j]) #removes empty plots
plt.tight_layout()



```

The majority of customers who do not churn tend to have sevices like PhoneService, InternetService(particularly OSL) and Online Security enabled. For sevices like OnlineBackup, TechSupport, and StreamingTV, churn rates are noticeably higher when these services are not used or are unavailable
